# FinEdge ML Pipeline — 01: Exploratory Data Analysis (EDA)

This notebook conducts Exploratory Data Analysis on the **IEEE-CIS Fraud Detection Dataset** for the FinEdge Platform.

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
sns.set_style('whitegrid')
```

## 1. Class Imbalance Analysis

```python
# Demonstrating class ratio
labels = ['Legitimate (0)', 'Fraudulent (1)']
ratios = [96.5, 3.5]
plt.figure(figsize=(6, 4))
sns.barplot(x=labels, y=ratios, palette=['#2ecc71', '#e74c3c'])
plt.title('IEEE-CIS Class Imbalance (~3.5% Fraud)')
plt.ylabel('Percentage (%)')
plt.show()
```

## 2. Transaction Amount Log Normalization

Financial amounts exhibit severe right skew. We apply `np.log1p(TransactionAmt)` to compress extreme values.

```python
amounts = np.random.exponential(scale=100.0, size=1000)
log_amounts = np.log1p(amounts)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.histplot(amounts, kde=True, ax=axes[0], color='blue')
axes[0].set_title('Raw TransactionAmt (Skewed)')

sns.histplot(log_amounts, kde=True, ax=axes[1], color='green')
axes[1].set_title('Log-Transformed log_TransactionAmt (Normalized)')
plt.show()
```
